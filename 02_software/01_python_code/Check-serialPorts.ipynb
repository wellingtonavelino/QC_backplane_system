```python
import serial
import serial.tools.list_ports

print("🔍 Scanning for active serial ports...\n")

for port in serial.tools.list_ports.comports():
    print(f"Checking {port.device} - {port.description}")
    try:
        with serial.Serial(port.device, baudrate=115200, timeout=1) as ser:
            data = ser.read(100)
            if data:
                print(f"✅ Data received from {port.device}: {data.decode(errors='replace')[:80]}")
            else:
                print(f"❌ No data received from {port.device}")
    except Exception as e:
        print(f"⚠️ Could not open {port.device}: {e}")

print("\n✔️ Done scanning.")
```
